## S5 · `bounds` — 함께 읽기

아래 함수를 **읽기만** 하세요. 아직 실행하지 않습니다.

**읽으면서 종이에 적을 것**

1. `lo` 와 `hi` 는 각각 **무엇을 담고 있는 이름**인가?
2. `if` 가 둘 있습니다. 각각 **어떤 변수를 읽고, 어떤 변수를 바꾸는가?**
3. 두 `if` 는 서로 **영향을 주는가?** 그렇게 판단한 근거는?


````python
def bounds(A):
    lo = A[0]
    hi = A[0]
    for x in A:
        if x < lo:
            lo = x
        if x > hi:
            hi = x
    return lo, hi
````

### 예상 — 실행 전에 적습니다

`bounds([4, -2, 7, 1])` 를 부르면 무엇이 나올까요?

**한 걸음씩 추적표를 채우세요.** `x` 를 하나 처리한 **직후**의 `(lo, hi)` 입니다.

| 처리한 x | 4 | -2 | 7 | 1 |
|---|---|---|---|---|
| (lo, hi) | | | | |

다 적었으면 아래를 실행합니다.


````python
print(bounds([4, -2, 7, 1]))
````

### 한 줄을 바꾸어 검증하기

`lo` 와 `hi` 의 **시작값을 둘 다 `0`** 으로 바꾸면 어떻게 될까요?

먼저 종이에 적으세요.

1. `[4, -2, 7, 1]` 에서는 결과가 달라질까?
2. **달라지는 입력**을 하나 만들 수 있을까? (원소 두 개짜리로 충분합니다)

원본은 그대로 두고 **새 이름**으로 만들어 비교합니다.


````python
def bounds_zero(A):
    lo = 0
    hi = 0
    for x in A:
        if x < lo:
            lo = x
        if x > hi:
            hi = x
    return lo, hi


test = [2, 5]
print('원본  ', bounds(test))
print('변경본', bounds_zero(test))
````

## S2 · `longest_run` — 예상 · 추적 · 실행

**읽으면서 종이에 적을 것**

1. `cur` 와 `best` 는 각각 무엇을 담는 이름인가?
2. `cur = 0` 은 **언제** 실행되는가?
3. `best` 는 **줄어들 수 있는가?**


````python
def longest_run(A):
    best = 0
    cur = 0
    for x in A:
        if x > 0:
            cur += 1
            if cur > best:
                best = cur
        else:
            cur = 0
    return best
````

### 예상 — 실행 전에 적습니다

`longest_run([2, 3, 0, 4, 5, 6, -1])` 의 결과는?

**추적표를 채우세요.** `x` 하나를 처리한 **직후**의 `(cur, best)` 입니다.

| 처리한 x | 2 | 3 | 0 | 4 | 5 | 6 | -1 |
|---|---|---|---|---|---|---|---|
| (cur, best) | | | | | | | |

그리고 하나 더 — 이 리스트에서 **양수는 모두 몇 개**인가요? 답과 **같은가요, 다른가요?**


````python
A = [2, 3, 0, 4, 5, 6, -1]
print(longest_run(A))
````

### 한 줄을 바꾸어 검증하기

마지막 줄을 `return best` 대신 **`return cur`** 로 바꾸면 결과는?

먼저 예상을 적고, 왜 그렇게 되는지 **`-1` 을 처리한 순간**을 근거로 설명하세요.


````python
def longest_run_cur(A):
    best = 0
    cur = 0
    for x in A:
        if x > 0:
            cur += 1
            if cur > best:
                best = cur
        else:
            cur = 0
    return cur


A = [2, 3, 0, 4, 5, 6, -1]
print('원본(return best)  ', longest_run(A))
print('변경본(return cur) ', longest_run_cur(A))
````

## S3 · `tidy` — 혼자 해석하고 검증하기

이번에는 **각자** 합니다. 순서는 같습니다 — 읽고 → 예상하고 → 실행하고 → 한 줄 바꿔 보고 → 설명하기.

**읽으면서 적을 것**

1. `keep` 과 `drop` 은 각각 **무엇을 모으는가?** 둘의 **자료형**은?
2. `keep.sort()` 는 **반복 안**에 있는가, **밖**에 있는가? 그래서 몇 번 실행되는가?


````python
def tidy(words, n):
    keep = []
    drop = 0
    for w in words:
        if len(w) >= n:
            keep.append(w)
        else:
            drop += 1
    keep.sort()
    return keep, drop
````

### 예상 — 실행 전에 적습니다

`tidy(['pear', 'a', 'plum', 'kiwi', 'to'], 4)` 에 대해 **둘 다** 적으세요.

1. **정렬 직전** `keep` 에는 무엇이 들어 있는가? (순서까지)
2. 최종 반환값은?


````python
words = ['pear', 'a', 'plum', 'kiwi', 'to']
print(tidy(words, 4))
````

### 한 줄을 바꾸어 검증하기

마지막 줄을 **`return len(keep), drop`** 으로 바꾸면?

예상을 적은 뒤 실행하세요. 그리고 이렇게 물어보세요 —
**「목록을 담는 것」과 「개수를 세는 것」은 무엇이 다른가?**


````python
def tidy_len(words, n):
    keep = []
    drop = 0
    for w in words:
        if len(w) >= n:
            keep.append(w)
        else:
            drop += 1
    keep.sort()
    return len(keep), drop


words = ['pear', 'a', 'plum', 'kiwi', 'to']
print('원본   ', tidy(words, 4))
print('변경본 ', tidy_len(words, 4))
````

## 유형 9-1 · 초기화가 **어디에** 있는가

두 코드는 **한 줄의 위치만** 다릅니다. `total = 0` 이 바깥 반복 **밖**에 있느냐, **안**에 있느냐.

두 함수를 **따로** 정의해 비교합니다. 같은 이름을 쓰면 나중에 실행한 것이 앞의 것을 덮어써서
비교 자체가 어긋납니다.


````python
def sum_all(m):
    total = 0
    for row in m:
        for x in row:
            total += x
    return total


def sum_rows(m):
    for row in m:
        total = 0
        for x in row:
            total += x
    return total
````

### 예상 — 실행 전에 적습니다

`m = [[1, 2], [3, 4]]` 일 때

1. `sum_all(m)` 은?
2. `sum_rows(m)` 은? 그리고 그 값은 **무엇의 합**인가?


````python
m = [[1, 2], [3, 4]]
print('sum_all ', sum_all(m))
print('sum_rows', sum_rows(m))
````

## Data Flow Tracer 연결 — 이름 셋의 역할

화면: **https://gbshs.github.io/afterschool/crc/tracer.html**

**53걸음을 다 받아 적지 않습니다.** 이름 셋이 무엇을 담는지만 잡으면 됩니다.

| 이름 | 무엇을 담는가 |
|---|---|
| `q` | 앞으로 **꺼내 처리할** 노드 |
| `seen` | 이미 **발견해서** 다시 넣지 않도록 표시한 노드 |
| `came` | 그 노드를 **어디에서 발견했는지** 기록 |

**주의할 구분 하나** — `seen` 에 표시되는 시점은 **발견한 순간**이고,
번호가 붙는 시점은 **큐에서 꺼낸 순간**입니다. 둘은 다릅니다.


### 예상 — 화면을 보기 전에 적습니다

그래프는 이렇게 이어져 있습니다.

```
A → B, C      B → D, E      C → F
D → G         E → G         F → G
```

세 장면만 예상하세요.

1. `A` 의 이웃을 다 넣은 뒤 `q = [B, C]` 입니다. **다음에 꺼내는 것**은?
2. `G` 를 **다시 발견**했을 때, 왜 `q` 에 **또 넣지 않는가?**
3. `came[G]` 에는 **무엇이** 기록되는가? (BFS 일 때)

적었으면 화면에서 확인하고, 아래 셀로 한 번 더 확인합니다.


````python
from collections import deque


def search(G, start, mode='bfs'):
    q = deque([start])
    seen = {start}
    came = {}
    order = []
    while q:
        cur = q.popleft() if mode == 'bfs' else q.pop()
        order.append(cur)
        for nxt in G[cur]:
            if nxt not in seen:
                seen.add(nxt)
                came[nxt] = cur
                q.append(nxt)
    return order, came


G = {'A': ['B', 'C'], 'B': ['D', 'E'], 'C': ['F'],
     'D': ['G'], 'E': ['G'], 'F': ['G'], 'G': []}

order, came = search(G, 'A', 'bfs')
print('꺼낸 순서 ', ' → '.join(order))
print('came[G] =', came['G'])
````

### DFS 와 견주기

위 코드에서 바뀌는 것은 **한 군데뿐**입니다 — `q.popleft()` 가 `q.pop()` 이 됩니다.

**예상을 먼저 적으세요.**

1. 꺼내는 순서는 어떻게 달라질까?
2. `came[G]` 는 그대로일까, 달라질까?


````python
order_d, came_d = search(G, 'A', 'dfs')
print('BFS 순서', ' → '.join(order))
print('DFS 순서', ' → '.join(order_d))
print('came[G]  BFS:', came['G'], '/ DFS:', came_d['G'])
````

## (선택) S6 · `rle` — 빨리 끝난 사람만

**필수가 아닙니다.** 앞의 것을 다 끝내고 시간이 남을 때만 하세요.


````python
def rle(s):
    out = []
    prev = ''
    cnt = 0
    for ch in s:
        if ch == prev:
            cnt += 1
        else:
            if cnt > 0:
                out.append((prev, cnt))
            prev = ch
            cnt = 1
    out.append((prev, cnt))
    return out
````

### 예상 — 실행 전에

1. `rle('aaabbc')` 는?
2. `rle('aba')` 는? — **글자별 개수**를 세는 것과 결과가 같은가, 다른가?
3. `rle('')` 는? (빈 문자열입니다)


````python
print(rle('aaabbc'))
print(rle('aba'))
print(rle(''))
````

### 한 줄을 바꾸어 검증하기

**반복 밖의 마지막 `out.append(...)` 를 지우면** 어떻게 될까요?
`rle('aaabbc')` 로 예상을 적고 확인하세요.


````python
def rle_nolast(s):
    out = []
    prev = ''
    cnt = 0
    for ch in s:
        if ch == prev:
            cnt += 1
        else:
            if cnt > 0:
                out.append((prev, cnt))
            prev = ch
            cnt = 1
    return out


print('원본   ', rle('aaabbc'))
print('변경본 ', rle_nolast('aaabbc'))
````

### 도전 — 빈 입력을 고치기

`rle('')` 이 `[('', 0)]` 대신 **`[]`** 가 나오게 하려면 어디를 어떻게 고쳐야 할까요?
고친 함수를 **새 이름**으로 만들고, `'aaabbc'` 에서도 결과가 그대로인지 확인하세요.


````python
def rle_fixed(s):
    out = []
    prev = ''
    cnt = 0
    for ch in s:
        if ch == prev:
            cnt += 1
        else:
            if cnt > 0:
                out.append((prev, cnt))
            prev = ch
            cnt = 1
    # 여기를 고쳐 보세요
    out.append((prev, cnt))
    return out


print(rle_fixed(''))
print(rle_fixed('aaabbc'))
````
